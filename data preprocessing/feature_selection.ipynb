```python
import pandas as pd
import numpy as np

df = pd.read_csv('FYP_Cleaned_Dataset.csv', parse_dates=['Datetime'])
print(f"Dataset loaded successfully with {df.shape[0]} rows and {df.shape[1]} columns.")
```

```python
import numpy as np

target = 'GHI_W_m2'

daylight_df = df[df['Is_Daylight'] == 1].copy()
numeric_df = daylight_df.select_dtypes(include=[np.number]).copy()

corr = numeric_df.corr()

ghi_corr = corr[target].sort_values(ascending=False)

print(f"Correlation with target: {target}")
print(ghi_corr)
```

```python
df.drop(columns=['Surface_Pressure_kPa'], inplace=True)
```

```python
df = df.sort_values(['State', 'Datetime']).reset_index(drop=True)

df['GHI_lag_1h'] = df.groupby('State')['GHI_W_m2'].shift(1)
df['GHI_lag_24h'] = df.groupby('State')['GHI_W_m2'].shift(24)
df = df.dropna(subset=['GHI_lag_1h', 'GHI_lag_24h']).reset_index(drop=True)
```

```python
df_cleaned = df.copy()
df_cleaned.to_csv('FYP_Cleaned_Dataset.csv', index=False)
```
