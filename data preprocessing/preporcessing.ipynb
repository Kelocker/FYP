```python
import pandas as pd
import numpy as np

df = pd.read_csv('malaysia_solar_hourly_analytics_prepared.csv', parse_dates=['Datetime'])
print(f"Dataset loaded successfully with {df.shape[0]} rows and {df.shape[1]} columns.")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Get a sorted list of your 14 states
states = sorted(df['State'].unique())

fig, axes = plt.subplots(4, 4, figsize=(16, 16), subplot_kw={'projection': 'polar'})
axes = axes.flatten()

# Loop through each state and draw its compass
for i, state in enumerate(states):
    # Filter data for the specific state
    state_data = df[df['State'] == state]['Wind_Direction_10m_deg'].dropna()
    
    # Convert degrees to radians (required for polar plots)
    radians = np.deg2rad(state_data)
    
    # Create the polar histogram (16 bins = 1 compass point every 22.5 degrees)
    bins = np.linspace(0.0, 2 * np.pi, 16, endpoint=False)
    axes[i].hist(radians, bins=bins, color='teal', edgecolor='black', alpha=0.7)
    
    axes[i].set_title(state, fontsize=12, fontweight='bold', pad=15)
    axes[i].set_theta_zero_location('N') 
    axes[i].set_theta_direction(-1)       
    
    # Set the compass labels
    axes[i].set_xticks(np.pi/180. * np.linspace(0, 360, 8, endpoint=False))
    axes[i].set_xticklabels(['N', 'NE', 'E', 'SE', 'S', 'SW', 'W', 'NW'])
    axes[i].set_yticklabels([])           


for j in range(i + 1, len(axes)):
    axes[j].axis('off')

plt.suptitle('Polar Distribution of Wind Direction by State', fontsize=20, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create a dataset of ONLY hours where it actually rained
positive_rain_df = df[df['Rainfall_mm'] > 0].copy()

# Set up a side-by-side figure
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot 1: Histogram of ALL data (shows the massive zero-inflation)
sns.histplot(df['Rainfall_mm'], bins=50, color='lightskyblue', ax=axes[0])
axes[0].set_title('Distribution of All Rainfall (Including Zeros)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Frequency (Hours)')
axes[0].set_xlabel('Rainfall (mm)')

# Plot 2: Boxplot of positive rain events
sns.boxplot(x='State', y='Rainfall_mm', data=positive_rain_df, color='royalblue', fliersize=3, ax=axes[1])
axes[1].set_title('Rainfall Outliers (Only When Rainfall > 0)', fontsize=12, fontweight='bold')
plt.setp(axes[1].get_xticklabels(), rotation=45, ha='right')
axes[1].set_ylabel('Rainfall (mm)')
axes[1].set_xlabel('State')

plt.suptitle('Rainfall Distribution and Event Outliers', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
df['Hour_sin'] = np.sin(2 * np.pi * df['Hour'] / 24)
df['Hour_cos'] = np.cos(2 * np.pi * df['Hour'] / 24)

df['Wind_Sine'] = np.sin(np.deg2rad(df['Wind_Direction_10m_deg']))
df['Wind_Cosine'] = np.cos(np.deg2rad(df['Wind_Direction_10m_deg']))

df_cleaned = df.drop(columns=['Hour', 'Wind_Direction_10m_deg'])

display(df_cleaned[['Datetime', 'Hour_sin', 'Hour_cos', 'Wind_Sine', 'Wind_Cosine']].head(24))
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Apply log1p transformation to handle skewed features containing zeros
df['Rainfall_mm_log'] = np.log1p(df['Rainfall_mm'])

# plot the before-and-after for Rainfall so you can visually prove 
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Plot 1: Original (Using a log scale on the Y-axis just to make the outliers visible)
sns.histplot(df['Rainfall_mm'], bins=50, ax=axes[0], color='skyblue')
axes[0].set_title('Original Rainfall (Highly Skewed)')
axes[0].set_yscale('log') 
axes[0].set_ylabel('Count (Log Scale)')

# Plot 2: Transformed
sns.histplot(df['Rainfall_mm_log'], bins=50, ax=axes[1], color='lightgreen')
axes[1].set_title('Log-Transformed Rainfall')
axes[1].set_ylabel('Count')

plt.tight_layout()
plt.show()

df.drop(columns=['Rainfall_mm'], inplace=True)
```
